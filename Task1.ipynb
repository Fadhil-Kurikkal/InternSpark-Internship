```python
encoder = LabelEncoder()
df['species'] = encoder.fit_transform(df['species'])
df.head()
```

```python
x = df.drop('species', axis=1)
y = df['species']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size=0.2, random_state=42, stratify=y
)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(x_train, y_train)
knn_pred = knn_model.predict(x_test)
knn_accuracy = accuracy_score(y_test, knn_pred)
print("Accuracy: ",knn_accuracy)
```

```python
LR_model = LogisticRegression(max_iter=200)
LR_model.fit(x_train, y_train)
LR_pred = LR_model.predict(x_test)
LR_accuracy = accuracy_score(y_test, LR_pred)
print("Accuracy: ",LR_accuracy)
```

```python
DT_model = DecisionTreeClassifier(random_state=42)
DT_model.fit(x_train, y_train)
DT_pred = DT_model.predict(x_test)
DT_accuracy = accuracy_score(y_test, DT_pred)
print("Accuracy: ",DT_accuracy)
```

```python
result = pd.DataFrame({
    'Model': ['knn_model', 'LR_model', 'DT_model'],
    'Accuracy': [knn_accuracy,LR_accuracy,DT_accuracy]
    })
result = result.sort_values(by='Accuracy', ascending=False)
result
```

```python
best_model_name = result.loc[
    result['Accuracy'].idxmax(), 'Model'
]
print("Best Model: ", best_model_name)
```

```python
best_model = knn_model
y_pred = best_model.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", conf_matrix)
```

```python
species_names = encoder.classes_

print(
    classification_report(
        y_test,
        y_pred,
        target_names=species_names
    )
)
```

```python
joblib.dump(
    best_model,
    "iris_model.pkl"
)

print("Model saved successfully.")
```

```python
joblib.dump(
    scaler,
    "scaler.pkl"
)

print("Scaler saved successfully.")
```
